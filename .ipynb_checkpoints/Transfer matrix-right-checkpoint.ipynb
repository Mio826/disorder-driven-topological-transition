# Project configuration & File I/O and metadata protection

```python
# ============================================================
# Cell 2
# Project configuration
#
# This is the main cell to edit for a new calculation.
# ============================================================

DEMO_MODE = False


# ------------------------------------------------------------
# Project and output labels
# ------------------------------------------------------------

PROJECT_NAME = "qwz_transfer_stage1"
BOUNDARY_LABEL = "left_impurity_boundary"

OUT_DIR = Path(PROJECT_NAME)
OUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# QWZ Hamiltonian
# ------------------------------------------------------------

MODEL_PARAMS = {
    "t": 1.0,
    "u": -2.05,
    "A": 1.5,
}

DENSITY = 0.03
FERMI_ENERGY = 0.0


# ------------------------------------------------------------
# Transfer-matrix finite-size scan
#
# Geometry:
#     x direction: long disordered strip
#     y direction: periodic transverse direction of width M
#
# The transfer backend uses independent Bernoulli disorder
# on every strip site.
# ------------------------------------------------------------

if DEMO_MODE:

    # These widths and strip lengths test the full pipeline only.
    WIDTHS = [3, 4, 5, 6]

    H_VALUES = np.linspace(
        -3.52,
        -3.38,
        8,
    )

    N_SLICES = 1200
    N_TRANSFER_SEEDS = 3

    TRANSFER_BOOTSTRAP = 20
    TRANSFER_POLYNOMIAL_ORDER = 2

else:

    # Replace these with production values after runtime benchmarks.
    WIDTHS = [20, 30, 40, 50]

    # H_VALUES = np.linspace(
    #     -3.48,
    #     -3.41,
    #     15,
    # )
    H_VALUES = np.array([
    -3.456,
    -3.454,
    -3.452,
    -3.450,
    -3.448,
    -3.446,
    -3.444,])

    N_SLICES = 100_000
    N_TRANSFER_SEEDS = 4

    TRANSFER_BOOTSTRAP = 500
    TRANSFER_POLYNOMIAL_ORDER = 3


TRANSFER_SEED0 = 20260730
QR_INTERVAL = 1


# ------------------------------------------------------------
# Transfer fit settings
# ------------------------------------------------------------

TRANSFER_FIT_H_WINDOW = (
    float(H_VALUES.min()),
    float(H_VALUES.max()),
)

MIN_TRANSFER_WIDTH = min(WIDTHS)

TRANSFER_NU_BOUNDS = (
    0.2,
    8.0,
)


# ------------------------------------------------------------
# Spectral and wavefunction scan at transfer-derived h_c
#
# These 2D diagonalizations do NOT come from the transfer
# matrix itself. They are companion measurements used to
# extract z and multifractal exponents at the same h_c.
# ------------------------------------------------------------

if DEMO_MODE:
    SPECTRAL_L_VALUES = [8, 10, 12]
    N_SPECTRAL_REALIZATIONS = 6
    DOS_BOOTSTRAP = 20
    MULTIFRACTAL_BOOTSTRAP = 20
else:
    SPECTRAL_L_VALUES = [16, 20, 24, 28, 32]
    N_SPECTRAL_REALIZATIONS = 100
    DOS_BOOTSTRAP = 500
    MULTIFRACTAL_BOOTSTRAP = 500


SPECTRAL_SEED0 = 20260731

# Keep this as "bernoulli" to match the transfer ensemble.
# "fixed_count" is available for a deliberate comparison.
SPECTRAL_ENSEMBLE = "bernoulli"
SPECTRAL_ROUNDING = "round"


# Critical-point selection:
#
#   "fit"    -> use h_c from the transfer fit
#   "manual" -> use HC_MANUAL
#
# Use "fit" for production only after the transfer fit is stable.
HC_SOURCE = "fit"
HC_MANUAL = -3.445


# ------------------------------------------------------------
# DOS settings
# ------------------------------------------------------------

DOS_BROADENING = 0.035

DOS_ENERGY_GRID = np.geomspace(
    0.02,
    0.35,
    36,
)

DOS_FIT_WINDOW = (
    0.06,
    0.25,
)


# ------------------------------------------------------------
# Multifractal settings
# ------------------------------------------------------------

Q_VALUES = np.array(
    [
        0.5,
        1.5,
        2.0,
        2.5,
        3.0,
    ],
    dtype=float,
)

MULTIFRACTAL_BOX_SIZE = 1


# ------------------------------------------------------------
# Run/load switches
# ------------------------------------------------------------

RUN_TRANSFER_SIMULATION = True
RUN_SPECTRAL_SIMULATION = True

ALLOW_METADATA_MISMATCH = False


# ------------------------------------------------------------
# Show configuration
# ------------------------------------------------------------

print("Boundary:", BOUNDARY_LABEL)
print("Transfer widths:", WIDTHS)
print("h range:", (float(H_VALUES.min()), float(H_VALUES.max())))
print("Slices per transfer realization:", N_SLICES)
print("Transfer seeds per point:", N_TRANSFER_SEEDS)
print("Spectral sizes:", SPECTRAL_L_VALUES)
print("Spectral realizations per size:", N_SPECTRAL_REALIZATIONS)
print("Density:", DENSITY)
print("Transfer ensemble: Bernoulli")
print("Spectral ensemble:", SPECTRAL_ENSEMBLE)

print("\nImportant:")
print(
    "DEMO_MODE=True tests the complete pipeline only; "
    "its fitted exponents are not production physics."
)


# ============================================================
# Cell 3
# File I/O and metadata protection
# ============================================================

TRANSFER_RAW_CSV = OUT_DIR / "transfer_raw.csv"
TRANSFER_SUMMARY_CSV = OUT_DIR / "transfer_summary.csv"

SPECTRUM_RAW_CSV = OUT_DIR / "spectrum_raw.csv"
STATE_RAW_CSV = OUT_DIR / "critical_state_raw.csv"
MULTIFRACTAL_RAW_CSV = OUT_DIR / "multifractal_raw.csv"

DOS_RAW_CSV = OUT_DIR / "dos_raw.csv"
DOS_SUMMARY_CSV = OUT_DIR / "dos_summary.csv"

RUN_METADATA_JSON = OUT_DIR / "run_metadata.json"


RUN_CONFIG = {
    "project_name": PROJECT_NAME,
    "boundary_label": BOUNDARY_LABEL,
    "model_params": MODEL_PARAMS,
    "density": float(DENSITY),
    "fermi_energy": float(FERMI_ENERGY),
    "widths": [int(M) for M in WIDTHS],
    "h_values": [float(h) for h in H_VALUES],
    "n_slices": int(N_SLICES),
    "n_transfer_seeds": int(N_TRANSFER_SEEDS),
    "transfer_seed0": int(TRANSFER_SEED0),
    "qr_interval": int(QR_INTERVAL),
    "transfer_fit_h_window": [
        float(TRANSFER_FIT_H_WINDOW[0]),
        float(TRANSFER_FIT_H_WINDOW[1]),
    ],
    "spectral_L_values": [
        int(L)
        for L in SPECTRAL_L_VALUES
    ],
    "n_spectral_realizations": int(N_SPECTRAL_REALIZATIONS),
    "spectral_seed0": int(SPECTRAL_SEED0),
    "spectral_ensemble": SPECTRAL_ENSEMBLE,
    "spectral_rounding": SPECTRAL_ROUNDING,
    "dos_broadening": float(DOS_BROADENING),
    "dos_energy_grid": [
        float(x)
        for x in DOS_ENERGY_GRID
    ],
    "q_values": [
        float(q)
        for q in Q_VALUES
    ],
    "multifractal_box_size": int(MULTIFRACTAL_BOX_SIZE),
}


if RUN_METADATA_JSON.exists():

    with RUN_METADATA_JSON.open(
        "r",
        encoding="utf-8",
    ) as handle:
        old_config = json.load(handle)

    if (
        old_config != RUN_CONFIG
        and not ALLOW_METADATA_MISMATCH
    ):
        raise RuntimeError(
            "The output folder already contains data generated with "
            "a different configuration.\n\n"
            "Use a new PROJECT_NAME, delete the old output folder, "
            "or explicitly set ALLOW_METADATA_MISMATCH=True."
        )

else:

    with RUN_METADATA_JSON.open(
        "w",
        encoding="utf-8",
    ) as handle:
        json.dump(
            RUN_CONFIG,
            handle,
            indent=2,
            sort_keys=True,
        )


print("Output directory:", OUT_DIR.resolve())

for path in [
    TRANSFER_RAW_CSV,
    TRANSFER_SUMMARY_CSV,
    SPECTRUM_RAW_CSV,
    STATE_RAW_CSV,
    MULTIFRACTAL_RAW_CSV,
    DOS_RAW_CSV,
    DOS_SUMMARY_CSV,
]:
    print(path.name)
```

# Setup & run

```python
# ============================================================
# Cell 4
# Construct the actual QWZ transfer model
# ============================================================

template_model = QWZModel(
    Lx=2,
    Ly=2,
    pbc=True,
    **MODEL_PARAMS,
)


print("Model parameters")
print("t =", template_model.t)
print("u =", template_model.u)
print("A =", template_model.A)

print("\nHamiltonian blocks shared by all backends")
print("onsite =")
print(template_model.onsite_block)

print("\nhopping_x =")
print(template_model.hopping_x)

print("\nhopping_y =")
print(template_model.hopping_y)

print("\nimpurity block at h=-3.45 =")
print(template_model.impurity_block(-3.45))


# ------------------------------------------------------------
# Local transfer-matrix consistency check
# ------------------------------------------------------------

test_width = max(WIDTHS)

test_transfer_model = fs.transfer_model_from_template(
    template_model,
    width=test_width,
)

test_workspace = fs.prepare_transfer_workspace(
    test_transfer_model,
    energy=FERMI_ENERGY,
)

test_rng = np.random.default_rng(
    20260732
)

test_mask = fs.random_slice_impurity_mask(
    test_width,
    density=DENSITY,
    rng=test_rng,
)

test_h = float(
    H_VALUES[len(H_VALUES) // 2]
)

test_slice = fs.build_slice_hamiltonian(
    test_transfer_model,
    test_mask,
    h=test_h,
)

slice_hermiticity_residual = float(
    np.max(
        np.abs(
            test_slice
            - test_slice.conj().T
        )
    )
)

psi_previous = (
    test_rng.normal(
        size=test_workspace.slice_dim
    )
    + 1.0j
    * test_rng.normal(
        size=test_workspace.slice_dim
    )
)

psi_current = (
    test_rng.normal(
        size=test_workspace.slice_dim
    )
    + 1.0j
    * test_rng.normal(
        size=test_workspace.slice_dim
    )
)

recurrence_residual = fs.transfer_recurrence_residual(
    test_workspace,
    test_mask,
    h=test_h,
    psi_previous=psi_previous,
    psi_current=psi_current,
)


print("\nTransfer geometry")
print("test width M =", test_width)
print("slice dimension =", test_workspace.slice_dim)
print("transfer dimension =", test_workspace.transfer_dim)

print("\nRandom test slice impurity mask:")
print(test_mask.astype(int))

print(
    "Slice Hermiticity residual:",
    f"{slice_hermiticity_residual:.3e}",
)

print(
    "Schrodinger recurrence residual:",
    f"{recurrence_residual:.3e}",
)


if slice_hermiticity_residual > 1.0e-10:
    raise AssertionError(
        "The transverse slice is not Hermitian."
    )

if recurrence_residual > 1.0e-10:
    raise AssertionError(
        "The transfer recurrence is inconsistent "
        "with the QWZ Schrodinger equation."
    )


# ============================================================
# Cell 5
# Run or load the checkpointable transfer scan
# ============================================================

start_time = time.perf_counter()


if RUN_TRANSFER_SIMULATION:

    transfer_raw, transfer_summary = fs.run_transfer_scan(
        widths=WIDTHS,
        h_values=H_VALUES,
        template=template_model,
        density=DENSITY,
        n_slices=N_SLICES,
        n_seeds=N_TRANSFER_SEEDS,
        seed0=TRANSFER_SEED0,
        energy=FERMI_ENERGY,
        qr_interval=QR_INTERVAL,
        raw_csv=TRANSFER_RAW_CSV,
        summary_csv=TRANSFER_SUMMARY_CSV,
        boundary_label=BOUNDARY_LABEL,
        verbose=True,
    )

else:

    if (
        not TRANSFER_RAW_CSV.exists()
        or not TRANSFER_SUMMARY_CSV.exists()
    ):
        raise FileNotFoundError(
            "RUN_TRANSFER_SIMULATION=False, but transfer CSV files "
            "do not exist."
        )

    transfer_raw = pd.read_csv(
        TRANSFER_RAW_CSV
    )

    transfer_summary = pd.read_csv(
        TRANSFER_SUMMARY_CSV
    )


elapsed = time.perf_counter() - start_time

print()
print(
    f"Transfer stage finished in {elapsed:.2f} s"
)

print(
    f"Raw rows: {len(transfer_raw)}"
)

print(
    f"Summary rows: {len(transfer_summary)}"
)

display(
    transfer_summary.sort_values(
        ["M", "h"]
    )
)


# ============================================================
# Cell 6
# Transfer raw-data quality control and visualization
# ============================================================

expected_raw_rows = (
    len(WIDTHS)
    * len(H_VALUES)
    * N_TRANSFER_SEEDS
)

duplicate_count = int(
    transfer_raw.duplicated(
        subset=[
            "M",
            "h",
            "seed",
        ]
    ).sum()
)

group_counts = (
    transfer_raw
    .groupby(["M", "h"])
    .size()
    .rename("n_rows")
    .reset_index()
)

incomplete_groups = group_counts[
    group_counts["n_rows"]
    != N_TRANSFER_SEEDS
]


print(
    "Expected raw rows:",
    expected_raw_rows,
)

print(
    "Actual raw rows:",
    len(transfer_raw),
)

print(
    "Duplicate realization rows:",
    duplicate_count,
)

print(
    "Incomplete (M,h) groups:",
    len(incomplete_groups),
)

print(
    "Maximum Lyapunov pairing residual:",
    transfer_raw[
        "lyapunov_pairing_residual"
    ].max(),
)


if duplicate_count != 0:
    raise RuntimeError(
        "Duplicate transfer realization rows detected."
    )

if not incomplete_groups.empty:
    display(incomplete_groups)
    warnings.warn(
        "Some transfer scan points are incomplete."
    )


valid_transfer = (
    np.isfinite(
        transfer_raw["Lambda"]
    )
    & (
        transfer_raw["Lambda"]
        > 0.0
    )
    & np.isfinite(
        transfer_raw["gamma_min_pos"]
    )
    & (
        transfer_raw["gamma_min_pos"]
        > 0.0
    )
)

if not np.all(valid_transfer):
    display(
        transfer_raw[
            ~valid_transfer
        ]
    )
    warnings.warn(
        "Some transfer realizations have invalid localization data."
    )


fig, axes = plt.subplots(
    1,
    2,
    figsize=(13.0, 4.6),
)


for M, group in transfer_summary.groupby("M"):

    group = group.sort_values("h")

    axes[0].errorbar(
        group["h"],
        group["mean_Lambda"],
        yerr=group["sem_Lambda"],
        marker="o",
        linewidth=1.5,
        capsize=2,
        label=f"M={int(M)}",
    )

    axes[1].plot(
        group["h"],
        group["max_pairing_residual"],
        marker="o",
        linewidth=1.5,
        label=f"M={int(M)}",
    )


axes[0].set_xlabel("h")
axes[0].set_ylabel(
    r"$\Lambda_M=\xi_M/M$"
)
axes[0].set_title(
    "Normalized localization length"
)
axes[0].legend(frameon=False)


axes[1].set_xlabel("h")
axes[1].set_ylabel(
    "maximum Lyapunov pairing residual"
)
axes[1].set_yscale("log")
axes[1].set_title(
    "Transfer numerical diagnostic"
)
axes[1].legend(frameon=False)


plt.tight_layout()
plt.show()
```

# Critical Exponent

```python
# ============================================================
# Cell 7
# Localization-length exponent nu from transfer matrix
# ============================================================

transfer_fit_config = ex.TransferNuConfig(
    width_col="M",
    h_col="h",
    raw_value_col="Lambda",

    min_width=MIN_TRANSFER_WIDTH,

    h_window=TRANSFER_FIT_H_WINDOW,

    polynomial_order=TRANSFER_POLYNOMIAL_ORDER,

    include_irrelevant=False,

    hc_bounds=TRANSFER_FIT_H_WINDOW,

    nu_bounds=TRANSFER_NU_BOUNDS,

    n_bootstrap=TRANSFER_BOOTSTRAP,
    confidence_level=0.95,
    seed=20260733,
)


transfer_fit_result = ex.fit_nu_from_transfer(
    transfer_raw,
    config=transfer_fit_config,
)


print("Primary transfer-matrix scaling result")
display(
    transfer_fit_result.summary_frame()
)

print(
    "Fit success:",
    transfer_fit_result.success,
)

print(
    "Optimizer message:",
    transfer_fit_result.message,
)

print(
    "chi2 / dof =",
    (
        transfer_fit_result.objective_value
        / transfer_fit_result.degrees_of_freedom
        if transfer_fit_result.degrees_of_freedom > 0
        else np.nan
    ),
)


# ------------------------------------------------------------
# Plot with consistent colors for each width
# ------------------------------------------------------------

def plot_transfer_result_consistent(
    result,
    *,
    figsize=(13.0, 4.6),
):
    data = result.predictions.copy()

    fig, axes = plt.subplots(
        1,
        2,
        figsize=figsize,
        dpi=150,
    )

    color_cycle = (
        plt.rcParams[
            "axes.prop_cycle"
        ]
        .by_key()["color"]
    )

    for color_index, (
        M,
        group,
    ) in enumerate(
        data.groupby("M")
    ):

        group = group.sort_values("h")

        color = color_cycle[
            color_index
            % len(color_cycle)
        ]

        axes[0].errorbar(
            group["h"],
            group["value"],
            yerr=group["error"],
            marker="o",
            linestyle="none",
            color=color,
            capsize=2,
            label=f"M={M:g}",
        )

        axes[0].plot(
            group["h"],
            group["value_fit"],
            color=color,
            linewidth=1.5,
        )

        collapsed = group.sort_values(
            "x_scaling"
        )

        axes[1].errorbar(
            collapsed["x_scaling"],
            collapsed["value"],
            yerr=collapsed["error"],
            marker="o",
            linestyle="none",
            color=color,
            capsize=2,
            label=f"M={M:g}",
        )

        axes[1].plot(
            collapsed["x_scaling"],
            collapsed["value_fit"],
            color=color,
            linewidth=1.5,
        )

    axes[0].set_xlabel("h")
    axes[0].set_ylabel(
        r"$\Lambda_M$"
    )
    axes[0].set_title(
        "Transfer-matrix scaling fit"
    )

    axes[1].set_xlabel(
        r"$(h-h_c)M^{1/\nu}$"
    )
    axes[1].set_ylabel(
        r"$\Lambda_M$"
    )
    axes[1].set_title(
        "Localization-length collapse"
    )

    for ax in axes:
        ax.legend(frameon=False)

    fig.tight_layout()

    return fig, axes


plot_transfer_result_consistent(
    transfer_fit_result
)

plt.show()


# ------------------------------------------------------------
# Minimal fit-stability diagnostic
# ------------------------------------------------------------

stability_rows = []

widths_sorted = sorted(
    int(M)
    for M in WIDTHS
)

for min_width in widths_sorted[:-1]:

    config = ex.TransferNuConfig(
        width_col="M",
        h_col="h",
        raw_value_col="Lambda",

        min_width=min_width,
        h_window=TRANSFER_FIT_H_WINDOW,

        polynomial_order=TRANSFER_POLYNOMIAL_ORDER,
        include_irrelevant=False,

        hc_bounds=TRANSFER_FIT_H_WINDOW,
        nu_bounds=TRANSFER_NU_BOUNDS,

        n_bootstrap=0,
        seed=20260734,
    )

    try:

        result = ex.fit_nu_from_transfer(
            transfer_raw,
            config=config,
        )

        stability_rows.append(
            {
                "min_width": min_width,
                "hc": result.parameters["hc"],
                "hc_error": result.standard_errors["hc"],
                "nu": result.parameters["nu"],
                "nu_error": result.standard_errors["nu"],
                "chi2_per_dof": (
                    result.objective_value
                    / result.degrees_of_freedom
                    if result.degrees_of_freedom > 0
                    else np.nan
                ),
                "success": result.success,
            }
        )

    except Exception as error:

        stability_rows.append(
            {
                "min_width": min_width,
                "hc": np.nan,
                "hc_error": np.nan,
                "nu": np.nan,
                "nu_error": np.nan,
                "chi2_per_dof": np.nan,
                "success": False,
                "error": str(error),
            }
        )


stability_table = pd.DataFrame(
    stability_rows
)

display(stability_table)


# ------------------------------------------------------------
# Save fit outputs
# ------------------------------------------------------------

transfer_fit_result.summary_frame().to_csv(
    OUT_DIR / "transfer_nu_fit_summary.csv",
    index=False,
)

transfer_fit_result.predictions.to_csv(
    OUT_DIR / "transfer_nu_predictions.csv",
    index=False,
)

transfer_fit_result.residuals.to_csv(
    OUT_DIR / "transfer_nu_residuals.csv",
    index=False,
)

stability_table.to_csv(
    OUT_DIR / "transfer_nu_stability.csv",
    index=False,
)

if transfer_fit_result.bootstrap_samples is not None:

    transfer_fit_result.bootstrap_samples.to_csv(
        OUT_DIR / "transfer_nu_bootstrap.csv",
        index=False,
    )


print()
print("Transfer estimate:")

print(
    f"hc = {transfer_fit_result.parameters['hc']:.8f}"
)

print(
    f"nu = {transfer_fit_result.parameters['nu']:.6f}"
)

if DEMO_MODE:
    print(
        "\nThis is a small-width pipeline result, "
        "not a physical universality estimate."
    )
```

# Dynamic Exponent

The transfer fit supplies the critical point. The dynamical exponent is then measured from a separate 2D spectral calculation.

```python
# ============================================================
# Cell 8
# Critical eigenspectrum and wavefunction moments
#
# The transfer fit supplies h_c and nu. The calculations below
# are separate 2D spectral/wavefunction measurements at that h_c:
#
#     eigenvalues  -> DOS -> z
#     eigenvectors -> P_q -> tau(q), D_q
# ============================================================

hc_fit = float(
    transfer_fit_result.parameters["hc"]
)


if HC_SOURCE == "manual":

    hc_spectral = float(
        HC_MANUAL
    )

elif HC_SOURCE == "fit":

    if (
        np.isfinite(hc_fit)
        and H_VALUES.min()
        <= hc_fit
        <= H_VALUES.max()
    ):
        hc_spectral = hc_fit

    else:
        warnings.warn(
            "The fitted transfer h_c lies outside the simulated h range. "
            "Using HC_MANUAL instead."
        )

        hc_spectral = float(
            HC_MANUAL
        )

else:

    raise ValueError(
        "HC_SOURCE must be 'fit' or 'manual'."
    )


print(
    f"Transfer fitted hc: {hc_fit:.8f}"
)

print(
    f"Spectral hc used:   {hc_spectral:.8f}"
)

print(
    "The z and multifractal calculations below are companion "
    "2D measurements, not outputs of the transfer matrix itself."
)


if RUN_SPECTRAL_SIMULATION:

    eigenvalue_rows = []
    state_rows = []
    multifractal_rows = []

    start_time = time.perf_counter()

    for size_index, L in enumerate(
        SPECTRAL_L_VALUES
    ):

        model = QWZModel(
            Lx=L,
            Ly=L,
            pbc=True,
            **MODEL_PARAMS,
        )

        for realization in range(
            N_SPECTRAL_REALIZATIONS
        ):

            seed = fs.make_realization_seed(
                SPECTRAL_SEED0,
                31,
                size_index,
                L,
                realization,
            )

            impurities = fs.sample_impurities(
                model,
                ensemble=SPECTRAL_ENSEMBLE,
                density=DENSITY,
                seed=seed,
                rounding=SPECTRAL_ROUNDING,
            )

            Hamiltonian = model.hamiltonian(
                impurities=impurities,
                h=hc_spectral,
                sparse=False,
            )

            eigenvalues, eigenvectors = (
                fs.diagonalize_hermitian(
                    Hamiltonian
                )
            )

            # ----------------------------------------------
            # Save the full finite-size spectrum
            # ----------------------------------------------

            for level_index, energy in enumerate(
                eigenvalues
            ):

                eigenvalue_rows.append(
                    {
                        "L": L,
                        "h": hc_spectral,
                        "realization": realization,
                        "seed": seed,
                        "level_index": level_index,
                        "E": float(energy),
                        "n_impurities": len(impurities),
                        "p_eff": (
                            len(impurities)
                            / model.n_sites
                        ),
                    }
                )

            # ----------------------------------------------
            # Choose the smallest positive-energy state
            # ----------------------------------------------

            positive_indices = np.flatnonzero(
                eigenvalues
                >= FERMI_ENERGY
            )

            if positive_indices.size == 0:
                raise RuntimeError(
                    "No nonnegative eigenvalue found."
                )

            state_index = int(
                positive_indices[
                    np.argmin(
                        eigenvalues[
                            positive_indices
                        ]
                        - FERMI_ENERGY
                    )
                ]
            )

            state_energy = float(
                eigenvalues[state_index]
            )

            state = eigenvectors[
                :,
                state_index,
            ]

            site_probability = (
                ex.site_probabilities(
                    state,
                    norb=model.norb,
                    normalize=True,
                )
            )

            impurity_site_indices = [
                x + L * y
                for x, y in impurities
            ]

            impurity_weight = float(
                np.sum(
                    site_probability[
                        impurity_site_indices
                    ]
                )
            ) if impurity_site_indices else 0.0

            # ----------------------------------------------
            # Generalized participation ratios
            # ----------------------------------------------

            gpr = (
                ex.generalized_participation_ratios(
                    site_probability,
                    q_values=Q_VALUES,
                    shape=(L, L),
                    box_size=MULTIFRACTAL_BOX_SIZE,
                )
            )

            for row in gpr.itertuples(
                index=False
            ):

                multifractal_rows.append(
                    {
                        "L": L,
                        "h": hc_spectral,
                        "realization": realization,
                        "seed": seed,
                        "E_state": state_energy,
                        "q": float(row.q),
                        "Pq": float(row.Pq),
                        "n_boxes": int(row.n_boxes),
                        "box_size": int(row.box_size),
                        "n_impurities": len(impurities),
                        "p_eff": (
                            len(impurities)
                            / model.n_sites
                        ),
                    }
                )

            ipr_row = gpr[
                np.isclose(
                    gpr["q"],
                    2.0,
                )
            ]

            ipr = (
                float(ipr_row["Pq"].iloc[0])
                if not ipr_row.empty
                else np.nan
            )

            state_rows.append(
                {
                    "L": L,
                    "h": hc_spectral,
                    "realization": realization,
                    "seed": seed,
                    "E_state": state_energy,
                    "abs_E_state": abs(
                        state_energy
                    ),
                    "IPR": ipr,
                    "PR": (
                        1.0 / ipr
                        if ipr > 0.0
                        else np.nan
                    ),
                    "impurity_weight": impurity_weight,
                    "n_impurities": len(impurities),
                    "p_eff": (
                        len(impurities)
                        / model.n_sites
                    ),
                }
            )

            print(
                f"[spectral] L={L}, "
                f"r={realization}, "
                f"E={state_energy:+.6e}, "
                f"IPR={ipr:.6e}"
            )

    spectrum_raw = pd.DataFrame(
        eigenvalue_rows
    )

    state_raw = pd.DataFrame(
        state_rows
    )

    multifractal_raw = pd.DataFrame(
        multifractal_rows
    )

    spectrum_raw.to_csv(
        SPECTRUM_RAW_CSV,
        index=False,
    )

    state_raw.to_csv(
        STATE_RAW_CSV,
        index=False,
    )

    multifractal_raw.to_csv(
        MULTIFRACTAL_RAW_CSV,
        index=False,
    )

    elapsed = (
        time.perf_counter()
        - start_time
    )

    print(
        f"\nSpectral stage finished in "
        f"{elapsed:.2f} s"
    )

else:

    required = [
        SPECTRUM_RAW_CSV,
        STATE_RAW_CSV,
        MULTIFRACTAL_RAW_CSV,
    ]

    missing = [
        path.name
        for path in required
        if not path.exists()
    ]

    if missing:
        raise FileNotFoundError(
            "Missing spectral output files:\n- "
            + "\n- ".join(missing)
        )

    spectrum_raw = pd.read_csv(
        SPECTRUM_RAW_CSV
    )

    state_raw = pd.read_csv(
        STATE_RAW_CSV
    )

    multifractal_raw = pd.read_csv(
        MULTIFRACTAL_RAW_CSV
    )


# display(state_raw)

# print(
#     "Spectrum rows:",
#     len(spectrum_raw),
# )

# print(
#     "Multifractal rows:",
#     len(multifractal_raw),
# )
# ============================================================
# Cell 9
# Disorder-averaged DOS and dynamical exponent z
#
# rho(E) is estimated with Gaussian broadening:
#
# rho(E) = 1/(L^2 sqrt(2pi) eta)
#          sum_n exp[-(E-E_n)^2/(2 eta^2)]
#
# The integral over all energies equals 2 states per lattice site.
# ============================================================

dos_rows = []

normalization_constant = (
    np.sqrt(2.0 * np.pi)
    * DOS_BROADENING
)


for (
    L,
    realization,
), group in spectrum_raw.groupby(
    [
        "L",
        "realization",
    ]
):

    eigenvalues = group[
        "E"
    ].to_numpy(dtype=float)

    for energy in DOS_ENERGY_GRID:

        kernel = np.exp(
            -0.5
            * (
                (
                    energy
                    - eigenvalues
                )
                / DOS_BROADENING
            ) ** 2
        )

        rho = float(
            np.sum(kernel)
            / (
                L**2
                * normalization_constant
            )
        )

        dos_rows.append(
            {
                "L": int(L),
                "realization": int(realization),
                "E": float(energy),
                "rho": rho,
            }
        )


dos_raw = pd.DataFrame(
    dos_rows
)

dos_summary = (
    dos_raw
    .groupby(
        [
            "L",
            "E",
        ],
        as_index=False,
    )
    .agg(
        n=("rho", "count"),
        rho=("rho", "mean"),
        rho_std=("rho", "std"),
    )
)

dos_summary["rho_sem"] = (
    dos_summary["rho_std"]
    / np.sqrt(
        dos_summary["n"]
    )
)


dos_raw.to_csv(
    DOS_RAW_CSV,
    index=False,
)

dos_summary.to_csv(
    DOS_SUMMARY_CSV,
    index=False,
)


# ------------------------------------------------------------
# Plot DOS for every size
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7.0, 5.0)
)

for L, group in dos_summary.groupby("L"):

    group = group.sort_values("E")

    ax.errorbar(
        group["E"],
        group["rho"],
        yerr=group["rho_sem"],
        marker="o",
        markersize=3.5,
        linewidth=1.2,
        capsize=2,
        label=f"L={int(L)}",
    )

ax.axvspan(
    DOS_FIT_WINDOW[0],
    DOS_FIT_WINDOW[1],
    alpha=0.12,
    label="fit window",
)

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlabel(r"$|E|$")
ax.set_ylabel(r"$\rho(E)$")
ax.set_title(
    rf"DOS near $h_c={hc_spectral:.6f}$"
)
ax.legend(frameon=False)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Fit z separately for each system size
# ------------------------------------------------------------

z_results = {}
z_rows = []

for L in sorted(
    dos_summary["L"].unique()
):

    size_dos = (
        dos_summary[
            dos_summary["L"] == L
        ]
        .copy()
    )

    z_config = ex.DOSZConfig(
        energy_col="E",
        dos_col="rho",
        error_col="rho_sem",
        dimension=2.0,
        energy_window=DOS_FIT_WINDOW,
        min_points=6,
        n_bootstrap=DOS_BOOTSTRAP,
        confidence_level=0.95,
        seed=20260728 + int(L),
    )

    try:

        result = (
            ex.fit_z_from_dos_powerlaw(
                size_dos,
                config=z_config,
            )
        )

        z_results[int(L)] = result

        z_rows.append(
            {
                "L": int(L),
                "z": result.parameters["z"],
                "z_error": (
                    result.standard_errors["z"]
                ),
                "alpha": (
                    result.parameters["alpha"]
                ),
                "amplitude": (
                    result.parameters[
                        "amplitude"
                    ]
                ),
                "success": result.success,
            }
        )

    except Exception as error:

        warnings.warn(
            f"DOS fit failed for L={L}: "
            f"{error}"
        )


z_table = pd.DataFrame(
    z_rows
)

display(z_table)


if z_results:

    largest_z_size = max(
        z_results
    )

    largest_z_result = (
        z_results[
            largest_z_size
        ]
    )

    print(
        f"Largest-size DOS result: "
        f"L={largest_z_size}"
    )

    display(
        largest_z_result.summary_frame()
    )

    ex.plot_dos_z_result(
        largest_z_result
    )

    plt.show()

    z_table.to_csv(
        OUT_DIR / "z_by_size.csv",
        index=False,
    )

else:

    largest_z_size = None
    largest_z_result = None

    warnings.warn(
        "No DOS fit succeeded."
    )


print(
    "\nImportant: this is a simple finite-L "
    "power-law estimate of z."
)

print(
    "A final result requires a stable energy window, "
    "size convergence, and tests of logarithmic corrections."
)
```

# Multifractal Spectrum

This section uses the same 2D critical eigenstates generated above.

```python
# ============================================================
# Cell 10
# Multifractal exponents tau(q), D_q, alpha and f(alpha)
# ============================================================

multifractal_config = (
    ex.MultifractalConfig(
        size_col="L",
        q_col="q",
        pq_col="Pq",
        realization_col="realization",

        min_size=min(
            SPECTRAL_L_VALUES
        ),

        q_values=tuple(
            float(q)
            for q in Q_VALUES
        ),

        compute_typical=True,
        compute_average=True,

        n_bootstrap=MULTIFRACTAL_BOOTSTRAP,
        confidence_level=0.95,
        seed=20260729,
    )
)


multifractal_result = (
    ex.fit_multifractal_spectrum(
        multifractal_raw,
        config=multifractal_config,
    )
)


print(
    "Multifractal fit success:",
    multifractal_result.success,
)

display(
    multifractal_result.spectrum
)

display(
    multifractal_result.scale_table
)


ex.plot_multifractal_result(
    multifractal_result
)

plt.show()


# ------------------------------------------------------------
# Direct IPR scaling diagnostic
# ------------------------------------------------------------

ipr_summary = (
    state_raw
    .groupby(
        "L",
        as_index=False,
    )
    .agg(
        n=("IPR", "count"),
        mean_IPR=("IPR", "mean"),
        mean_log_IPR=(
            "IPR",
            lambda values: float(
                np.mean(
                    np.log(values)
                )
            ),
        ),
        mean_impurity_weight=(
            "impurity_weight",
            "mean",
        ),
        mean_abs_E=(
            "abs_E_state",
            "mean",
        ),
    )
)

ipr_summary["typical_IPR"] = np.exp(
    ipr_summary[
        "mean_log_IPR"
    ]
)


display(ipr_summary)


fig, axes = plt.subplots(
    1,
    2,
    figsize=(12.5, 4.5),
)


axes[0].loglog(
    ipr_summary["L"],
    ipr_summary["typical_IPR"],
    marker="o",
    label="typical IPR",
)

axes[0].loglog(
    ipr_summary["L"],
    ipr_summary["mean_IPR"],
    marker="s",
    label="average IPR",
)

axes[0].set_xlabel("L")
axes[0].set_ylabel("IPR")
axes[0].set_title(
    r"Critical-state $P_2$ scaling"
)
axes[0].legend(frameon=False)


axes[1].plot(
    ipr_summary["L"],
    ipr_summary[
        "mean_impurity_weight"
    ],
    marker="o",
)

axes[1].set_xlabel("L")
axes[1].set_ylabel(
    "probability on impurity sites"
)

axes[1].set_title(
    "Microscopic impurity weight"
)


plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# Save multifractal outputs
# ------------------------------------------------------------

multifractal_result.spectrum.to_csv(
    OUT_DIR / "multifractal_spectrum.csv",
    index=False,
)

multifractal_result.scale_table.to_csv(
    OUT_DIR / "multifractal_scale_table.csv",
    index=False,
)

if (
    multifractal_result.bootstrap_samples
    is not None
):

    multifractal_result.bootstrap_samples.to_csv(
        OUT_DIR
        / "multifractal_bootstrap.csv",
        index=False,
    )


print()
print(
    "For the left boundary, box_size=1 still contains "
    "strong microscopic impurity geometry."
)

print(
    "A final universality analysis must test larger boxes "
    "and larger L."
)
```

```python
# ============================================================
# Cell 11
# Final fixed-point summary and export
# ============================================================

hc_value = float(
    transfer_fit_result.parameters["hc"]
)

nu_value = float(
    transfer_fit_result.parameters["nu"]
)

hc_error = float(
    transfer_fit_result.standard_errors.get(
        "hc",
        np.nan,
    )
)

nu_error = float(
    transfer_fit_result.standard_errors.get(
        "nu",
        np.nan,
    )
)


if largest_z_result is not None:

    z_value = float(
        largest_z_result.parameters["z"]
    )

    z_error = float(
        largest_z_result.standard_errors.get(
            "z",
            np.nan,
        )
    )

else:

    z_value = np.nan
    z_error = np.nan


q2_rows = (
    multifractal_result.spectrum[
        np.isclose(
            multifractal_result.spectrum["q"],
            2.0,
        )
    ]
)


if not q2_rows.empty:

    q2_row = q2_rows.iloc[0]

    D2_typical = float(
        q2_row.get(
            "Dq_typical",
            np.nan,
        )
    )

    D2_average = float(
        q2_row.get(
            "Dq_average",
            np.nan,
        )
    )

    D2_typical_error = float(
        q2_row.get(
            "Dq_typical_error",
            np.nan,
        )
    )

    D2_average_error = float(
        q2_row.get(
            "Dq_average_error",
            np.nan,
        )
    )

else:

    D2_typical = np.nan
    D2_average = np.nan
    D2_typical_error = np.nan
    D2_average_error = np.nan


final_summary = pd.DataFrame(
    [
        {
            "quantity": "hc_transfer",
            "estimate": hc_value,
            "uncertainty": hc_error,
            "source": "transfer-matrix scaling",
        },
        {
            "quantity": "nu_transfer",
            "estimate": nu_value,
            "uncertainty": nu_error,
            "source": "transfer-matrix scaling",
        },
        {
            "quantity": (
                f"z_DOS_L{largest_z_size}"
                if largest_z_size is not None
                else "z_DOS"
            ),
            "estimate": z_value,
            "uncertainty": z_error,
            "source": "2D DOS at transfer-derived hc",
        },
        {
            "quantity": "D2_typical",
            "estimate": D2_typical,
            "uncertainty": D2_typical_error,
            "source": "2D critical wavefunction scaling",
        },
        {
            "quantity": "D2_average",
            "estimate": D2_average,
            "uncertainty": D2_average_error,
            "source": "2D critical wavefunction scaling",
        },
    ]
)


display(final_summary)

final_summary.to_csv(
    OUT_DIR / "final_fixed_point_summary.csv",
    index=False,
)


json_summary = {
    "boundary": BOUNDARY_LABEL,
    "demo_mode": DEMO_MODE,
    "hc_spectral_used": float(hc_spectral),
    "hc_transfer": hc_value,
    "hc_transfer_error": hc_error,
    "nu_transfer": nu_value,
    "nu_transfer_error": nu_error,
    "z_largest_size": z_value,
    "z_error": z_error,
    "D2_typical": D2_typical,
    "D2_average": D2_average,
}


with (
    OUT_DIR
    / "final_fixed_point_summary.json"
).open(
    "w",
    encoding="utf-8",
) as handle:

    json.dump(
        json_summary,
        handle,
        indent=2,
        sort_keys=True,
    )


print()
print("=" * 72)
print("QWZ TRANSFER-ANCHORED FIXED-POINT PIPELINE FINISHED")
print("=" * 72)

print(
    f"Transfer hc       = {hc_value:.8f}"
)

print(
    f"Transfer nu       = {nu_value:.6f}"
)

print(
    f"DOS z             = {z_value:.6f}"
)

print(
    f"Typical D2        = {D2_typical:.6f}"
)

print(
    f"Average D2        = {D2_average:.6f}"
)

print()
print(
    "Output folder:",
    OUT_DIR.resolve(),
)


if DEMO_MODE:

    print()
    print(
        "DEMO MODE WARNING:"
    )

    print(
        "These exponents come from small widths, short strips, "
        "and small 2D systems. They validate the pipeline, "
        "not the asymptotic universality class."
    )
```
